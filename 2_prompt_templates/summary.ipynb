# Lesson 02 - Exploring LangChain Prompt Templates

Prompt templates help us build reusable prompts with variables. Instead of writing a new prompt every time, we write the structure once and fill in the changing values when the program runs.

In this lesson, we will learn four core ideas:

- **Template** - the reusable prompt structure
- **Variables** - placeholders like `{club_name}` that get filled later
- **Messages** - system and user prompt roles for chat models
- **Invocation** - using `.invoke({...})` to fill the template


## Setup

This notebook summarizes `1_prompt_templates_starter.py`. The example uses OpenAI through LangChain and loads the API key from `.env`.


````python
# Install the packages used in this folder
pip install langchain-openai python-dotenv
````

````python
from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate

load_dotenv()

llm = ChatOpenAI(model="gpt-5-nano")
````

## Understanding The Prompt Template Flow

The example follows this pattern:

```text
Template text -> ChatPromptTemplate -> invoke with values -> completed prompt -> llm.invoke() -> response.content
```

1. Write a reusable prompt with placeholders.
2. Turn it into a `ChatPromptTemplate`.
3. Fill the placeholders using `.invoke({...})`.
4. Send the completed prompt to the model.
5. Print the response content.

The most important beginner idea is that the prompt structure and the prompt values are separate.


## Example 1 - A Simple String Template

The first example in the file is commented out, but it is important for learning. It shows how to make one reusable prompt string.


````python
template = "Write a {tone} email to {recipient} about {topic}."

prompt_template = ChatPromptTemplate.from_template(template)

prompt = prompt_template.invoke({
    "tone": "formal",
    "recipient": "John",
    "topic": "the weather",
})

print(prompt)
````

### What The Important Lines Do

- `{tone}`, `{recipient}`, and `{topic}` are placeholders.
- `ChatPromptTemplate.from_template(template)` creates a prompt template from one string.
- `.invoke({...})` fills the placeholders using a dictionary.
- The dictionary keys must match the placeholder names exactly.

This makes the prompt reusable. You can change the values without rewriting the whole prompt.


## Example 2 - A Chat Prompt With System And User Messages

The active code in `1_prompt_templates_starter.py` uses `ChatPromptTemplate.from_messages(...)`. This is useful because chat models understand roles.


````python
messages = [
    ("system", "You are a helpful assistant that can answer questions about {club_name}."),
    ("user", "What year was {club_name} founded?"),
]

prompt_template = ChatPromptTemplate.from_messages(messages)

prompt = prompt_template.invoke({"club_name": "Arsenal"})
print(prompt)

res = llm.invoke(prompt)
print(res.content)
````

### Understanding The Chat Prompt

- The `system` message gives the assistant instructions.
- The `user` message contains the user's question.
- `{club_name}` appears in both messages, but it only needs to be provided once.
- `ChatPromptTemplate.from_messages(messages)` builds a structured chat prompt.
- `prompt_template.invoke({"club_name": "Arsenal"})` replaces `{club_name}` with `Arsenal`.
- `llm.invoke(prompt)` sends the completed prompt to the model.
- `res.content` prints the model's final answer.

This structure is clearer than one long string because instructions and user requests are separated.


## Why Prompt Templates Matter

Prompt templates help beginners avoid messy string building. They make it clear which parts of the prompt stay the same and which parts change.

For example, this template:

```text
What year was {club_name} founded?
```

can be reused for Arsenal, Chelsea, Real Madrid, Barcelona, or any other club.

Templates are especially useful when building larger AI apps because they make prompts easier to test, update, and reuse.
